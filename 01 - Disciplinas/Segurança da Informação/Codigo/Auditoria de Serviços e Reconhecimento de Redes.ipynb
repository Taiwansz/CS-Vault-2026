## Parte A: Pesquisa Inicial

**Questão 1: Qual a diferença entre um scan de Host Discovery e um Port Scanning?**

> **Host Discovery:** Identifica quais dispositivos estão ativos na rede (IPs vivos).
> **Port Scanning:** Identifica quais serviços/portas estão abertos em um host específico já descoberto.

**Questão 2: Por que o uso da função settimeout() é crítico para o desempenho de um scanner de portas?**

> Evita que o script espere tempo demais por uma resposta de portas fechadas ou filtradas, acelerando drasticamente a varredura ao definir um limite curto de espera.

# Atividade Prática: Reconhecimento e Port Scanning

```python
import socket
import time
from datetime import datetime

# 1. Configurações Iniciais
host = input('🎯 Informe o IP ou hostname do alvo: ').strip()
try:
    p_inicial = int(input('📌 Porta inicial: '))
    p_final = int(input('📌 Porta final: '))
except ValueError:
    print('❌ Erro: Portas devem ser números inteiros.')
    exit()

# 2. Lista para armazenar resultados completos (Parte C)
resultados_completos = []
inicio_scan = datetime.now()
inicio_ts = time.time()

print(f'\n🔍 Varrendo {host}... Aguarde.')

# 3. Execução do Scan (Lógica da Parte B estendida)
for porta in range(p_inicial, p_final + 1):
    sock = socket.socket(socket.AF_INET, socket.SOCK_STREAM)
    sock.settimeout(0.5)

    resultado = sock.connect_ex((host, porta))
    status = 'ABERTA' if resultado == 0 else 'FECHADA/FILTRADA'

    # Armazena o resultado de todas as portas
    resultados_completos.append({'porta': porta, 'status': status})

    if status == 'ABERTA':
        print(f'✅ Porta {porta}: {status}')

    sock.close()

duracao = time.time() - inicio_ts

# 4. Geração do Relatório TXT (Parte C)
nome_arquivo = 'relatorio_auditoria.txt'
with open(nome_arquivo, 'w', encoding='utf-8') as f:
    # 5. Cabeçalho com metadados
    f.write('====================================================\n')
    f.write('      RELATÓRIO DE AUDITORIA DE SEGURANÇA           \n')
    f.write('====================================================\n')
    f.write(f'Data/Hora: {inicio_scan.strftime("%d/%m/%Y %H:%M:%S")}\n')
    f.write(f'Alvo:      {host}\n')
    f.write(f'Intervalo: {p_inicial} - {p_final}\n')
    f.write('----------------------------------------------------\n\n')

    # 6. Listagem detalhada
    f.write('PORTA\t\tSTATUS\n')
    for res in resultados_completos:
        f.write(f"{res['porta']}\t\t{res['status']}\n")

    # 7. Tempo total
    f.write('\n----------------------------------------------------\n')
    f.write(f'Tempo total de execução: {duracao:.2f} segundos\n')
    f.write('====================================================\n')

# 8. Confirmação
print(f'\n📄 Relatório "{nome_arquivo}" gerado com sucesso!')
```
