```python
"""
1 - Sistema de Animais de Estimação:
"""

class AnimalEstimacao:
    def __init__(self, nome, idade, especie):
        self.nome = nome
        self.idade = idade
        self.especie = especie

    def __str__(self):
        return f"{self.nome} é um(a) {self.especie} com {self.idade} anos de idade."


class Cachorro(AnimalEstimacao):
    def __init__(self, nome, idade, raca):
        super().__init__(nome, idade, "Cachorro")
        self.raca = raca

    def __str__(self):
        return f"{super().__str__()} Ele é um(a) {self.raca}."


class Gato(AnimalEstimacao):
    def __init__(self, nome, idade, cor_pelo):
        super().__init__(nome, idade, "Gato")
        self.cor_pelo = cor_pelo

    def __str__(self):
        return f"{super().__str__()} Ele tem pelo {self.cor_pelo}."


class Passaro(AnimalEstimacao):
    def __init__(self, nome, idade, tipo):
        super().__init__(nome, idade, "Pássaro")
        self.tipo = tipo

    def __str__(self):
        return f"{super().__str__()} Ele é um(a) {self.tipo}."


# Exemplo de uso:
if __name__ == "__main__":
    rex = Cachorro("Rex", 3, "Labrador")
    print(rex)

    whiskers = Gato("Whiskers", 5, "cinza")
    print(whiskers)

    tweety = Passaro("Tweety", 1, "Canário")
    print(tweety)
```

```python
"""
2 - Sistema de Jogos de Tabuleiro
"""
class JogoTabuleiro:
    def __init__(self, nome):
        self.nome = nome
        self.jogadores = []

    def iniciar_jogo(self):
        print(f"O jogo de {self.nome} está começando!")

    def finalizar_jogo(self):
        print(f"O jogo de {self.nome} terminou!")

    def adicionar_jogador(self, jogador):
        self.jogadores.append(jogador)
        print(f"{jogador} foi adicionado ao jogo de {self.nome}.")


class Xadrez(JogoTabuleiro):
    def __init__(self):
        super().__init__("Xadrez")
        self.tabuleiro = self._criar_tabuleiro()

    def _criar_tabuleiro(self):
        return "Tabuleiro de xadrez criado"


class Damas(JogoTabuleiro):
    def __init__(self):
        super().__init__("Damas")
        self.tabuleiro = self._criar_tabuleiro()

    def _criar_tabuleiro(self):
        return "Tabuleiro de damas criado"


class BancoImobiliario(JogoTabuleiro):
    def __init__(self):
        super().__init__("Banco Imobiliário")
        self.tabuleiro = self._criar_tabuleiro()

    def _criar_tabuleiro(self):
        return "Tabuleiro do Banco Imobiliário criado"


if __name__ == "__main__":
    xadrez = Xadrez()
    xadrez.iniciar_jogo()
    xadrez.adicionar_jogador("Alice")
    xadrez.adicionar_jogador("Bob")
    xadrez.finalizar_jogo()

    damas = Damas()
    damas.iniciar_jogo()
    damas.adicionar_jogador("Charlie")
    damas.finalizar_jogo()

    banco_imobiliario = BancoImobiliario()
    banco_imobiliario.iniciar_jogo()
    banco_imobiliario.adicionar_jogador("David")
    banco_imobiliario.adicionar_jogador("Eve")
    banco_imobiliario.finalizar_jogo()
```

```python
"""
3 - Sistema de Funcionários Públicos
"""

class FuncionarioPublico:
    def __init__(self, nome, idade, cargo):
        self.nome = nome
        self.idade = idade
        self.cargo = cargo

    def __str__(self):
        return f"{self.cargo}: {self.nome}, {self.idade} anos"


class Policial(FuncionarioPublico):
    def __init__(self, nome, idade, cargo, departamento):
        super().__init__(nome, idade, cargo)
        self.departamento = departamento

    def prender_criminoso(self, criminoso):
        print(f"{self.nome} prendeu {criminoso}!")

    def __str__(self):
        return f"{super().__str__()}, Departamento: {self.departamento}"


class Bombeiro(FuncionarioPublico):
    def __init__(self, nome, idade, cargo, estacao):
        super().__init__(nome, idade, cargo)
        self.estacao = estacao

    def combater_incendio(self, local):
        print(f"{self.nome} está combatendo o incêndio em {local}!")

    def __str__(self):
        return f"{super().__str__()}, Estação: {self.estacao}"


class ProfissionalSaude(FuncionarioPublico):
    def __init__(self, nome, idade, cargo, especialidade):
        super().__init__(nome, idade, cargo)
        self.especialidade = especialidade

    def tratar_paciente(self, paciente):
        print(f"{self.nome} está tratando o paciente {paciente}!")

    def __str__(self):
        return f"{super().__str__()}, Especialidade: {self.especialidade}"



if __name__ == "__main__":
    policial = Policial("João", 35, "Policial", "Departamento de Polícia")
    print(policial)
    policial.prender_criminoso("Carlos")

    bombeiro = Bombeiro("Pedro", 28, "Bombeiro", "Estação Central")
    print(bombeiro)
    bombeiro.combater_incendio("Rua Principal")

    medico = ProfissionalSaude("Maria", 40, "Médico", "Cardiologia")
    print(medico)
    medico.tratar_paciente("Ana")
```

```python
"""
4 - Sistema de Documentos
"""

class Documento:
    def __init__(self, nome):
        self.nome = nome
        self.aberto = False

    def abrir(self):
        if not self.aberto:
            print(f"Abrindo o documento {self.nome}.")
            self.aberto = True
        else:
            print("O documento já está aberto.")

    def fechar(self):
        if self.aberto:
            print(f"Fechando o documento {self.nome}.")
            self.aberto = False
        else:
            print("O documento já está fechado.")

    def salvar(self):
        if self.aberto:
            print(f"Salvando as alterações no documento {self.nome}.")
        else:
            print("Não é possível salvar um documento fechado.")


class DocumentoTexto(Documento):
    def __init__(self, nome, conteudo=""):
        super().__init__(nome)
        self.conteudo = conteudo

    def editar(self, novo_conteudo):
        if self.aberto:
            print("Editando o documento de texto.")
            self.conteudo = novo_conteudo
        else:
            print("O documento precisa ser aberto para editar.")

    def __str__(self):
        return f"Documento de Texto: {self.nome}"


class Planilha(Documento):
    def __init__(self, nome, dados=None):
        super().__init__(nome)
        self.dados = dados if dados else {}

    def adicionar_dados(self, chave, valor):
        if self.aberto:
            print(f"Adicionando dados à planilha: {chave} - {valor}.")
            self.dados[chave] = valor
        else:
            print("A planilha precisa ser aberta para adicionar dados.")

    def __str__(self):
        return f"Planilha: {self.nome}"


class Apresentacao(Documento):
    def __init__(self, nome, slides=None):
        super().__init__(nome)
        self.slides = slides if slides else []

    def adicionar_slide(self, novo_slide):
        if self.aberto:
            print("Adicionando um novo slide à apresentação.")
            self.slides.append(novo_slide)
        else:
            print("A apresentação precisa ser aberta para adicionar slides.")

    def __str__(self):
        return f"Apresentação: {self.nome}"



if __name__ == "__main__":
    documento_texto = DocumentoTexto("MeuDocumento.txt")
    documento_texto.abrir()
    documento_texto.editar("Este é o conteúdo do meu documento de texto.")
    documento_texto.salvar()
    documento_texto.fechar()

    planilha = Planilha("MinhaPlanilha.xlsx")
    planilha.abrir()
    planilha.adicionar_dados("A1", 100)
    planilha.adicionar_dados("B2", 200)
    planilha.salvar()
    planilha.fechar()

    apresentacao = Apresentacao("MinhaApresentacao.pptx")
    apresentacao.abrir()
    apresentacao.adicionar_slide("Slide 1")
    apresentacao.adicionar_slide("Slide 2")
    apresentacao.salvar()
    apresentacao.fechar()
```

```python
"""
5 - Sistema de Produtos Eletrônicos
"""

class ProdutoEletronico:
    def __init__(self, nome, preco, marca):
        self.nome = nome
        self.preco = preco
        self.marca = marca

    def __str__(self):
        return f"{self.nome} - Marca: {self.marca}, Preço: R${self.preco:.2f}"


class Smartphone(ProdutoEletronico):
    def __init__(self, nome, preco, marca, sistema_operacional):
        super().__init__(nome, preco, marca)
        self.sistema_operacional = sistema_operacional

    def __str__(self):
        return f"Smartphone {super().__str__()}, Sistema Operacional: {self.sistema_operacional}"


class Tablet(ProdutoEletronico):
    def __init__(self, nome, preco, marca, tamanho_tela):
        super().__init__(nome, preco, marca)
        self.tamanho_tela = tamanho_tela

    def __str__(self):
        return f"Tablet {super().__str__()}, Tamanho da Tela: {self.tamanho_tela} polegadas"


class Notebook(ProdutoEletronico):
    def __init__(self, nome, preco, marca, processador):
        super().__init__(nome, preco, marca)
        self.processador = processador

    def __str__(self):
        return f"Notebook {super().__str__()}, Processador: {self.processador}"



if __name__ == "__main__":
    smartphone = Smartphone("iPhone 13", 6999.99, "Apple", "iOS")
    print(smartphone)

    tablet = Tablet("iPad Pro", 2399.99, "Apple", 12.9)
    print(tablet)

    notebook = Notebook("Dell XPS 15", 3299.99, "Dell", "Intel Core i7")
    print(notebook)
```

```python
"""
6 - Sistema de Veículos Automotores
"""

class VeiculoAutomotor:
    def __init__(self, marca, modelo):
        self.marca = marca
        self.modelo = modelo
        self.ligado = False
        self.velocidade = 0

    def ligar(self):
        if not self.ligado:
            print(f"{self.modelo} da marca {self.marca} ligado.")
            self.ligado = True
        else:
            print("O veículo já está ligado.")

    def desligar(self):
        if self.ligado:
            print(f"{self.modelo} da marca {self.marca} desligado.")
            self.ligado = False
            self.velocidade = 0
        else:
            print("O veículo já está desligado.")

    def acelerar(self, quantidade):
        if self.ligado:
            self.velocidade += quantidade
            print(f"{self.modelo} da marca {self.marca} acelerando. Velocidade atual: {self.velocidade} km/h")
        else:
            print("O veículo precisa estar ligado para acelerar.")


class Carro(VeiculoAutomotor):
    def __init__(self, marca, modelo, num_portas):
        super().__init__(marca, modelo)
        self.num_portas = num_portas

    def abrir_porta(self, num_porta):
        print(f"A porta {num_porta} do carro foi aberta.")

    def fechar_porta(self, num_porta):
        print(f"A porta {num_porta} do carro foi fechada.")


class Moto(VeiculoAutomotor):
    def __init__(self, marca, modelo, cilindrada):
        super().__init__(marca, modelo)
        self.cilindrada = cilindrada

    def buzinar(self):
        print("Biiiiiiiiiiiiiiiiiiiiiiiiiiiiiip!")


class Caminhao(VeiculoAutomotor):
    def __init__(self, marca, modelo, carga_maxima):
        super().__init__(marca, modelo)
        self.carga_maxima = carga_maxima

    def carregar(self):
        print(f"{self.modelo} da marca {self.marca} sendo carregado.")



if __name__ == "__main__":
    carro = Carro("Toyota", "Corolla", 4)
    carro.ligar()
    carro.acelerar(50)
    carro.abrir_porta(1)
    carro.fechar_porta(1)
    carro.desligar()

    moto = Moto("Honda", "CBR600RR", "600cc")
    moto.ligar()
    moto.acelerar(80)
    moto.buzinar()
    moto.desligar()

    caminhao = Caminhao("Volvo", "FH16", "40 toneladas")
    caminhao.ligar()
    caminhao.acelerar(30)
    caminhao.carregar()
    caminhao.desligar()
```

```python
"""
7 - Sistema de Espaçonaves
"""
class Espaconave:
    def __init__(self, nome):
        self.nome = nome
        self.coord_x = 0
        self.coord_y = 0
        self.coord_z = 0
        self.em_viagem = False

    def decolar(self):
        print(f"{self.nome} decolando...")

    def aterrissar(self):
        print(f"{self.nome} aterrissando...")

    def viajar_para(self, coord_x, coord_y, coord_z):
        if not self.em_viagem:
            self.coord_x = coord_x
            self.coord_y = coord_y
            self.coord_z = coord_z
            print(f"{self.nome} viajando para as coordenadas ({coord_x}, {coord_y}, {coord_z})...")
            self.em_viagem = True
        else:
            print(f"{self.nome} já está em viagem.")

    def __str__(self):
        return f"{self.nome} está atualmente em ({self.coord_x}, {self.coord_y}, {self.coord_z})."


class NaveTripulada(Espaconave):
    def __init__(self, nome, tripulantes):
        super().__init__(nome)
        self.tripulantes = tripulantes

    def informar_tripulantes(self):
        print(f"A {self.nome} possui {self.tripulantes} tripulantes.")


class SondaEspacial(Espaconave):
    def __init__(self, nome, instrumentos):
        super().__init__(nome)
        self.instrumentos = instrumentos

    def informar_instrumentos(self):
        print(f"A {self.nome} possui os seguintes instrumentos: {', '.join(self.instrumentos)}.")


class EstacaoEspacial(Espaconave):
    def __init__(self, nome, capacidade):
        super().__init__(nome)
        self.capacidade = capacidade

    def informar_capacidade(self):
        print(f"A {self.nome} tem capacidade para abrigar {self.capacidade} pessoas.")



if __name__ == "__main__":
    nave_tripulada = NaveTripulada("Nave Apollo", 3)
    nave_tripulada.decolar()
    nave_tripulada.viajar_para(100, 200, 300)
    nave_tripulada.aterrissar()
    print(nave_tripulada)
    nave_tripulada.informar_tripulantes()

    sonda_espacial = SondaEspacial("Voyager 1", ["Câmera", "Espectrômetro", "Plasma Analyzer"])
    sonda_espacial.decolar()
    sonda_espacial.viajar_para(500, 600, 700)
    sonda_espacial.aterrissar()
    print(sonda_espacial)
    sonda_espacial.informar_instrumentos()

    estacao_espacial = EstacaoEspacial("ISS", 10)
    estacao_espacial.decolar()
    estacao_espacial.viajar_para(1000, 1500, 2000)
    estacao_espacial.aterrissar()
    print(estacao_espacial)
    estacao_espacial.informar_capacidade()
```

```python
"""
8 - Sistema de Instrumentos Musicais
"""
class InstrumentoMusical:
    def __init__(self, nome):
        self.nome = nome
        self.tocando = False

    def tocar(self):
        if not self.tocando:
            print(f"Tocando o {self.nome}.")
            self.tocando = True
        else:
            print("O instrumento já está sendo tocado.")

    def parar(self):
        if self.tocando:
            print(f"Parando de tocar o {self.nome}.")
            self.tocando = False
        else:
            print("O instrumento já está parado.")

    def afinar(self):
        print(f"Afinando o {self.nome}.")


class Violao(InstrumentoMusical):
    def __init__(self):
        super().__init__("Violão")

    def trocar_corda(self):
        print("Trocando corda do violão.")


class Violino(InstrumentoMusical):
    def __init__(self):
        super().__init__("Violino")

    def mudar_arco(self):
        print("Mudando o arco do violino.")


class Bateria(InstrumentoMusical):
    def __init__(self):
        super().__init__("Bateria")

    def afinar_pratos(self):
        print("Afinando os pratos da bateria.")



if __name__ == "__main__":
    violao = Violao()
    violao.tocar()
    violao.trocar_corda()
    violao.parar()

    violino = Violino()
    violino.tocar()
    violino.mudar_arco()
    violino.parar()

    bateria = Bateria()
    bateria.tocar()
    bateria.afinar_pratos()
    bateria.parar()
```

```python
"""
9 - Sistema de Plantas
"""
class Planta:
    def __init__(self, nome, tipo, altura):
        self.nome = nome
        self.tipo = tipo
        self.altura = altura

    def crescer(self, crescimento):
        self.altura += crescimento
        print(f"A planta {self.nome} cresceu {crescimento} centímetros.")

    def __str__(self):
        return f"{self.nome} ({self.tipo}) - Altura: {self.altura} cm"


class Arvore(Planta):
    def __init__(self, nome, altura, frutifera=False):
        super().__init__(nome, "Árvore", altura)
        self.frutifera = frutifera

    def dar_frutos(self):
        if self.frutifera:
            print(f"A árvore {self.nome} está dando frutos.")
        else:
            print(f"A árvore {self.nome} não é frutífera.")


class Flor(Planta):
    def __init__(self, nome, altura, cor):
        super().__init__(nome, "Flor", altura)
        self.cor = cor

    def desabrochar(self):
        print(f"A flor {self.nome} está desabrochando na cor {self.cor}.")


class PlantaAquatica(Planta):
    def __init__(self, nome, altura, profundidade_maxima):
        super().__init__(nome, "Planta Aquática", altura)
        self.profundidade_maxima = profundidade_maxima

    def flutuar(self):
        print(f"A planta aquática {self.nome} está flutuando.")


if __name__ == "__main__":
    arvore = Arvore("Carvalho", 200, True)
    arvore.crescer(10)
    arvore.dar_frutos()
    print(arvore)

    flor = Flor("Rosa", 30, "vermelha")
    flor.desabrochar()
    print(flor)

    planta_aquatica = PlantaAquatica("Alga", 5, 50)
    planta_aquatica.flutuar()
    print(planta_aquatica)
```

```python
"""
10 - Sistema de Produtos de Limpeza
"""

class ProdutoLimpeza:
    def __init__(self, nome, preco, finalidade):
        self.nome = nome
        self.preco = preco
        self.finalidade = finalidade

    def __str__(self):
        return f"{self.nome} - Preço: R${self.preco:.2f}, Finalidade: {self.finalidade}"


class Detergente(ProdutoLimpeza):
    def __init__(self, nome, preco, fragrancia):
        super().__init__(nome, preco, "Limpeza de louças")
        self.fragrancia = fragrancia

    def __str__(self):
        return f"Detergente {super().__str__()}, Fragrância: {self.fragrancia}"


class Desinfetante(ProdutoLimpeza):
    def __init__(self, nome, preco, tipo_superficie):
        super().__init__(nome, preco, "Desinfecção")
        self.tipo_superficie = tipo_superficie

    def __str__(self):
        return f"Desinfetante {super().__str__()}, Tipo de Superfície: {self.tipo_superficie}"


class Amaciante(ProdutoLimpeza):
    def __init__(self, nome, preco, tipo_tecido):
        super().__init__(nome, preco, "Amaciamento de tecidos")
        self.tipo_tecido = tipo_tecido

    def __str__(self):
        return f"Amaciante {super().__str__()}, Tipo de Tecido: {self.tipo_tecido}"



if __name__ == "__main__":
    detergente = Detergente("Detergente Neutro", 5.99, "Lavanda")
    print(detergente)

    desinfetante = Desinfetante("Desinfetante de Pinho", 7.99, "Pisos")
    print(desinfetante)

    amaciante = Amaciante("Amaciante de Roupas Floral", 9.99, "Algodão")
    print(amaciante)
```
