```python
"""
1 - Calculadora Polimórfica
"""
class Calculadora:
    def calcular(self, operacao, *operandos):
        if operacao == "soma":
            return self.soma(*operandos)
        elif operacao == "subtracao":
            return self.subtracao(*operandos)
        elif operacao == "multiplicacao":
            return self.multiplicacao(*operandos)
        elif operacao == "divisao":
            return self.divisao(*operandos)
        else:
            raise ValueError("Operação não suportada")

    def soma(self, *operandos):
        resultado = operandos[0]
        for operando in operandos[1:]:
            resultado += operando
        return resultado

    def subtracao(self, *operandos):
        resultado = operandos[0]
        for operando in operandos[1:]:
            resultado -= operando
        return resultado

    def multiplicacao(self, *operandos):
        resultado = operandos[0]
        for operando in operandos[1:]:
            resultado *= operando
        return resultado

    def divisao(self, *operandos):
        resultado = operandos[0]
        for operando in operandos[1:]:
            resultado /= operando
        return resultado



if __name__ == "__main__":
    calculadora = Calculadora()

    resultado_soma = calculadora.calcular("soma", 5, 3, 2)
    print("Resultado da soma:", resultado_soma)

    resultado_subtracao = calculadora.calcular("subtracao", 10, 5, 3)
    print("Resultado da subtração:", resultado_subtracao)

    resultado_multiplicacao = calculadora.calcular("multiplicacao", 2, 3, 4)
    print("Resultado da multiplicação:", resultado_multiplicacao)

    resultado_divisao = calculadora.calcular("divisao", 100, 2, 5)
    print("Resultado da divisão:", resultado_divisao)
```

```python
"""
2 - Loja Virtual
"""
class Produto:
    def __init__(self, nome, preco):
        self.nome = nome
        self.preco = preco

    def calcular_preco(self):
        return self.preco


class Livro(Produto):
    def __init__(self, nome, preco, autor):
        super().__init__(nome, preco)
        self.autor = autor

    def calcular_preco(self):
        return self.preco * 0.9


class Eletronico(Produto):
    def __init__(self, nome, preco, garantia_meses):
        super().__init__(nome, preco)
        self.garantia_meses = garantia_meses

    def calcular_preco(self):
        if self.garantia_meses > 12:
            return self.preco * 0.95
        else:
            return self.preco


class Roupa(Produto):
    def __init__(self, nome, preco, tamanho):
        super().__init__(nome, preco)
        self.tamanho = tamanho

    def calcular_preco(self):
        if self.tamanho == "P":
            return self.preco * 0.9
        else:
            return self.preco



if __name__ == "__main__":
    livro = Livro("Dom Casmurro", 39.99, "Machado de Assis")
    print("Preço do livro:", livro.calcular_preco())

    eletronico = Eletronico("Smartphone", 1999.99, 24)
    print("Preço do eletrônico:", eletronico.calcular_preco())

    roupa = Roupa("Camiseta", 29.99, "M")
    print("Preço da roupa:", roupa.calcular_preco())
```

```python
"""
3 - Sistema Bancário
"""
class ContaBancaria:
    def __init__(self, titular, saldo=0):
        self.titular = titular
        self.saldo = saldo

    def depositar(self, valor):
        self.saldo += valor
        print(f"Depósito de R${valor} realizado. Novo saldo: R${self.saldo:.2f}")

    def sacar(self, valor):
        if self.saldo >= valor:
            self.saldo -= valor
            print(f"Saque de R${valor} realizado. Novo saldo: R${self.saldo:.2f}")
        else:
            print("Saldo insuficiente.")

    def calcular_juros(self):
        pass

    def calcular_taxas(self):
        pass


class ContaCorrente(ContaBancaria):
    def calcular_taxas(self):
        taxa = 5
        print(f"Taxa de manutenção da conta corrente: R${taxa:.2f}")
        self.saldo -= taxa


class ContaPoupanca(ContaBancaria):
    def calcular_juros(self):
        taxa_juros = 0.03
        juros = self.saldo * taxa_juros
        print(f"Juros da conta poupança: R${juros:.2f}")
        self.saldo += juros


class ContaInvestimento(ContaBancaria):
    def calcular_juros(self):
        taxa_juros = 0.05
        juros = self.saldo * taxa_juros
        print(f"Juros da conta investimento: R${juros:.2f}")
        self.saldo += juros



if __name__ == "__main__":
    conta_corrente = ContaCorrente("João", 1000)
    conta_corrente.depositar(500)
    conta_corrente.calcular_taxas()
    print("Saldo atual da conta corrente:", conta_corrente.saldo)

    conta_poupanca = ContaPoupanca("Maria", 2000)
    conta_poupanca.depositar(1000)
    conta_poupanca.calcular_juros()
    print("Saldo atual da conta poupança:", conta_poupanca.saldo)

    conta_investimento = ContaInvestimento("Pedro", 3000)
    conta_investimento.depositar(1500)
    conta_investimento.calcular_juros()
    print("Saldo atual da conta investimento:", conta_investimento.saldo)
```

```python
"""
6 - Gestão de Veículos
"""
class Veiculo:
    def __init__(self, marca, modelo, ano, preco):
        self.marca = marca
        self.modelo = modelo
        self.ano = ano
        self.preco = preco

    def calcular_consumo_combustivel(self):
        pass

    def calcular_custo_manutencao(self):
        pass


class Carro(Veiculo):
    def __init__(self, marca, modelo, ano, preco, consumo_medio):
        super().__init__(marca, modelo, ano, preco)
        self.consumo_medio = consumo_medio

    def calcular_consumo_combustivel(self, distancia):
        return distancia / self.consumo_medio

    def calcular_custo_manutencao(self):
        return self.preco * 0.03


class Moto(Veiculo):
    def __init__(self, marca, modelo, ano, preco, cilindrada):
        super().__init__(marca, modelo, ano, preco)
        self.cilindrada = cilindrada

    def calcular_consumo_combustivel(self, distancia):
        return distancia / (self.cilindrada / 200)

    def calcular_custo_manutencao(self):
        return self.preco * 0.02


class Caminhao(Veiculo):
    def __init__(self, marca, modelo, ano, preco, capacidade_carga):
        super().__init__(marca, modelo, ano, preco)
        self.capacidade_carga = capacidade_carga

    def calcular_consumo_combustivel(self, distancia):
        return distancia / (self.capacidade_carga / 1000)

    def calcular_custo_manutencao(self):
        return self.preco * 0.05



if __name__ == "__main__":
    carro = Carro("Toyota", "Corolla", 2020, 60000, 12)
    print("Consumo de combustível do carro para 100 km:", carro.calcular_consumo_combustivel(100))
    print("Custo de manutenção do carro:", carro.calcular_custo_manutencao())

    moto = Moto("Honda", "CBR 1000RR", 2019, 30000, 1000)
    print("Consumo de combustível da moto para 100 km:", moto.calcular_consumo_combustivel(100))
    print("Custo de manutenção da moto:", moto.calcular_custo_manutencao())

    caminhao = Caminhao("Volvo", "FH16", 2021, 200000, 20000)
    print("Consumo de combustível do caminhão para 100 km:", caminhao.calcular_consumo_combustivel(100))
    print("Custo de manutenção do caminhão:", caminhao.calcular_custo_manutencao())
```

```python
"""
7 - Sistema de Empréstimos
"""
class Emprestimo:
    def __init__(self, valor, taxa_juros):
        self.valor = valor
        self.taxa_juros = taxa_juros

    def calcular_juros(self, meses):
        return self.valor * self.taxa_juros * meses

    def calcular_prazo_pagamento(self, valor_parcela):
        return self.valor / valor_parcela


class EmprestimoPessoal(Emprestimo):
    def __init__(self, valor, taxa_juros):
        super().__init__(valor, taxa_juros)

    def calcular_juros(self, meses):
        return super().calcular_juros(meses)

    def calcular_prazo_pagamento(self, valor_parcela):
        return min(36, super().calcular_prazo_pagamento(valor_parcela))


class EmprestimoEmpresarial(Emprestimo):
    def __init__(self, valor, taxa_juros):
        super().__init__(valor, taxa_juros)

    def calcular_juros(self, meses):
        return super().calcular_juros(meses) * 0.8

    def calcular_prazo_pagamento(self, valor_parcela):
        return min(60, super().calcular_prazo_pagamento(valor_parcela))


class EmprestimoEstudantil(Emprestimo):
    def __init__(self, valor, taxa_juros):
        super().__init__(valor, taxa_juros)

    def calcular_juros(self, meses):
        return super().calcular_juros(meses) * 0.5

    def calcular_prazo_pagamento(self, valor_parcela):
        return min(120, super().calcular_prazo_pagamento(valor_parcela))



if __name__ == "__main__":
    emprestimo_pessoal = EmprestimoPessoal(10000, 0.02)
    juros_pessoal = emprestimo_pessoal.calcular_juros(12)
    prazo_pessoal = emprestimo_pessoal.calcular_prazo_pagamento(500)
    print("Juros do empréstimo pessoal:", juros_pessoal)
    print("Prazo de pagamento do empréstimo pessoal:", prazo_pessoal)

    emprestimo_empresarial = EmprestimoEmpresarial(50000, 0.015)
    juros_empresarial = emprestimo_empresarial.calcular_juros(24)
    prazo_empresarial = emprestimo_empresarial.calcular_prazo_pagamento(2000)
    print("Juros do empréstimo empresarial:", juros_empresarial)
    print("Prazo de pagamento do empréstimo empresarial:", prazo_empresarial)

    emprestimo_estudantil = EmprestimoEstudantil(20000, 0.01)
    juros_estudantil = emprestimo_estudantil.calcular_juros(36)
    prazo_estudantil = emprestimo_estudantil.calcular_prazo_pagamento(300)
    print("Juros do empréstimo estudantil:", juros_estudantil)
    print("Prazo de pagamento do empréstimo estudantil:", prazo_estudantil)
```

```python
"""
13 - Gestão de Equipamentos Esportivos
"""
class EquipamentoEsportivo:
    def __init__(self, nome, preco_base):
        self.nome = nome
        self.preco_base = preco_base

    def calcular_preco_aluguel(self, dias):
        pass

    def verificar_disponibilidade(self):
        pass


class Bola(EquipamentoEsportivo):
    def __init__(self, nome, preco_base):
        super().__init__(nome, preco_base)

    def calcular_preco_aluguel(self, dias):
        return self.preco_base * dias

    def verificar_disponibilidade(self):
        return True


class Raquete(EquipamentoEsportivo):
    def __init__(self, nome, preco_base):
        super().__init__(nome, preco_base)

    def calcular_preco_aluguel(self, dias):
        return self.preco_base + 5 * dias

    def verificar_disponibilidade(self):
        estoque_disponivel = True
        return estoque_disponivel


class Luvas(EquipamentoEsportivo):
    def __init__(self, nome, preco_base):
        super().__init__(nome, preco_base)

    def calcular_preco_aluguel(self, dias):
        return self.preco_base + 7 * dias

    def verificar_disponibilidade(self):
        estoque_disponivel = True
        return estoque_disponivel



if __name__ == "__main__":
    bola = Bola("Bola de Futebol", 10)
    preco_aluguel_bola = bola.calcular_preco_aluguel(3)  # Aluguel por 3 dias
    disponibilidade_bola = bola.verificar_disponibilidade()
    print("Preço de aluguel da bola:", preco_aluguel_bola)
    print("Disponibilidade da bola para aluguel:", disponibilidade_bola)

    raquete = Raquete("Raquete de Tênis", 20)
    preco_aluguel_raquete = raquete.calcular_preco_aluguel(2)  # Aluguel por 2 dias
    disponibilidade_raquete = raquete.verificar_disponibilidade()
    print("Preço de aluguel da raquete:", preco_aluguel_raquete)
    print("Disponibilidade da raquete para aluguel:", disponibilidade_raquete)

    luvas = Luvas("Luvas de Boxe", 15)
    preco_aluguel_luvas = luvas.calcular_preco_aluguel(4)  # Aluguel por 4 dias
    disponibilidade_luvas = luvas.verificar_disponibilidade()
    print("Preço de aluguel das luvas:", preco_aluguel_luvas)
    print("Disponibilidade das luvas para aluguel:", disponibilidade_luvas)
```
