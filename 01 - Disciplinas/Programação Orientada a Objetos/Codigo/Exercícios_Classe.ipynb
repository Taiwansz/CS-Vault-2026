```python
'''Exercicio 1 - Crie uma classe chamada Retangulo que tenha métodos para calcular a área
e o perímetro do retângulo, dados os valores dos lados.'''

class Retangulo:
    def __init__(self, comprimento, largura):
        self.comprimento = comprimento
        self.largura = largura

    def calcular_area(self):
        return self.comprimento * self.largura

    def calcular_perimetro(self):
        return 2 * (self.comprimento + self.largura)

# Exemplo de uso
retangulo = Retangulo(5, 3)
print("Área do retângulo:", retangulo.calcular_area())
print("Perímetro do retângulo:", retangulo.calcular_perimetro())
```

```python
'''Exercício 2'''

class ContaBancaria:
    def __init__(self, saldo=0):
        self.saldo = saldo

    def depositar(self, valor):
        self.saldo += valor
        return self.saldo

    def sacar(self, valor):
        if self.saldo >= valor:
            self.saldo -= valor
            return self.saldo
        else:
            print("Saldo insuficiente")
            return None

    def verificar_saldo(self):
        return self.saldo

    def transferir(self, outra_conta, valor):
        if self.saldo >= valor:
            self.saldo -= valor
            outra_conta.depositar(valor)
            print("Transferência realizada com sucesso.")
        else:
            print("Saldo insuficiente para transferência.")

    def __str__(self):
        return f"Saldo atual: {self.saldo}"


if __name__ == "__main__":
    conta1 = ContaBancaria(1000)
    conta2 = ContaBancaria(500)

    print("Saldo da conta 1:", conta1.verificar_saldo())
    print("Saldo da conta 2:", conta2.verificar_saldo())

    conta1.transferir(conta2, 300)

    print("Saldo da conta 1 após transferência:", conta1.verificar_saldo())
    print("Saldo da conta 2 após transferência:", conta2.verificar_saldo())
```

```python
'''Exercício 3'''

class Pessoa:
    def __init__(self, nome, idade, genero):
        self.nome = nome
        self.idade = idade
        self.genero = genero

    def atualizar_nome(self, novo_nome):
        self.nome = novo_nome

    def atualizar_idade(self, nova_idade):
        self.idade = nova_idade

    def atualizar_genero(self, novo_genero):
        self.genero = novo_genero

    def exibir_informacoes(self):
        return f"Nome: {self.nome}, Idade: {self.idade}, Gênero: {self.genero}"


if __name__ == "__main__":
    pessoa1 = Pessoa("João", 30, "Macho")
    print(pessoa1.exibir_informacoes())

    pessoa1.atualizar_idade(31)
    pessoa1.atualizar_genero("Femea")
    print(pessoa1.exibir_informacoes())
```

```python
'''Exercício 4'''

class Livro:
    def __init__(self, titulo, autor, paginas):
        self.titulo = titulo
        self.autor = autor
        self.paginas = paginas
        self.emprestado = False  # Inicialmente o livro não está emprestado

    def exibir_informacoes(self):
        print(f"Título: {self.titulo}")
        print(f"Autor: {self.autor}")
        print(f"Número de páginas: {self.paginas}")
        print("Status: ", "Disponível" if not self.emprestado else "Emprestado")

    def emprestar(self):
        if not self.emprestado:
            self.emprestado = True
            print("Livro emprestado com sucesso.")
        else:
            print("Livro já está emprestado.")

    def devolver(self):
        if self.emprestado:
            self.emprestado = False
            print("Livro devolvido com sucesso.")
        else:
            print("Livro já está disponível.")

    def verificar_disponibilidade(self):
        return not self.emprestado

if __name__ == "__main__":
    livro1 = Livro("O Senhor dos Anéis", "J.R.R. Tolkien", 1000)
    livro1.exibir_informacoes()
    print()

    livro1.emprestar()
    livro1.exibir_informacoes()
    print()

    livro1.devolver()
    livro1.exibir_informacoes()
```

```python
'''Exercício 5'''

import math

class Triangulo:
    def __init__(self, lado1, lado2, lado3):
        self.lado1 = lado1
        self.lado2 = lado2
        self.lado3 = lado3

    def calcular_area(self):
        perimetro = self.calcular_perimetro() # corrigido
        area = math.sqrt(perimetro/2 * (perimetro/2 - self.lado1) * (perimetro/2 - self.lado2) * (perimetro/2 - self.lado3))
        return area

    def calcular_perimetro(self):
        return self.lado1 + self.lado2 + self.lado3

    def tipo_triangulo(self):
        if self.lado1 == self.lado2 == self.lado3:
            return "Equilátero"
        elif self.lado1 == self.lado2 or self.lado1 == self.lado3 or self.lado2 == self.lado3:
            return "Isósceles"
        else:
            return "Escaleno"

# Exemplo de uso:
if __name__ == "__main__":
    triangulo1 = Triangulo(3, 3, 3)
    print("Tipo de Triângulo:", triangulo1.tipo_triangulo())

    triangulo2 = Triangulo(3, 4, 4)
    print("Tipo de Triângulo:", triangulo2.tipo_triangulo())

    triangulo3 = Triangulo(3, 4, 5)
    print("Tipo de Triângulo:", triangulo3.tipo_triangulo())
```

```python
'''Exercício 7'''

import datetime

class Calendario:
    def __init__(self):
        self.data_atual = datetime.date.today()

    def verificar_ano_bissexto(self):
        if self.data_atual.year % 4 == 0:
            if self.data_atual.year % 100 == 0:
                if self.data_atual.year % 400 == 0:
                    return True
                else:
                    return False
            else:
                return True
        else:
            return False

    def exibir_dia_semana(self):
        dias_semana = ["Segunda-feira", "Terça-feira", "Quarta-feira", "Quinta-feira", "Sexta-feira", "Sábado", "Domingo"]
        return dias_semana[self.data_atual.weekday()]

    def __str__(self):
        return f"Data atual: {self.data_atual}"

# Exemplo de uso:
if __name__ == "__main__":
    calendario = Calendario()
    print("Ano bissexto:", calendario.verificar_ano_bissexto())
    print("Dia da semana:", calendario.exibir_dia_semana())
    print(calendario)
```

```python
'''Exercício 8'''

class Estudante:
    def __init__(self, nome, idade, nota, serie):
        self.nome = nome
        self.idade = idade
        self.nota = nota
        self.serie = serie

    def atualizar_nota(self, nova_nota):
        self.nota = nova_nota

    def atualizar_serie(self, nova_serie):
        self.serie = nova_serie

    def __str__(self):
        return f"Nome: {self.nome}, Idade: {self.idade}, Nota: {self.nota}, Série: {self.serie}"

# Exemplo de uso:
if __name__ == "__main__":
    estudante1 = Estudante("João", 16, 8.5, "9º ano")
    print(estudante1)

    estudante1.atualizar_nota(9.0)
    estudante1.atualizar_serie("3º ano")
    print(estudante1)
```

```python
'''Exercício 9'''

class Funcionario:
    def __init__(self, nome, cargo, salario):
        self.nome = nome
        self.cargo = cargo
        self.salario = salario

    def calcular_aumento_salario(self, porcentagem):
        self.salario *= (1 + porcentagem/100)
        return self.salario

    def __str__(self):
        return f"Nome: {self.nome}, Cargo: {self.cargo}, Salário: R${self.salario:.2f}"

# Exemplo de uso:
if __name__ == "__main__":
    funcionario1 = Funcionario("João", "Gari", 1400.0)
    print(funcionario1)

    funcionario1.calcular_aumento_salario(10)
    print(funcionario1)
```

```python
'''Exercício 10'''

class Produto:
    def __init__(self, nome, preco, quantidade_estoque):
        self.nome = nome
        self.preco = preco
        self.quantidade_estoque = quantidade_estoque

    def atualizar_preco(self, novo_preco):
        self.preco = novo_preco

    def atualizar_quantidade_estoque(self, nova_quantidade):
        self.quantidade_estoque = nova_quantidade

    def exibir_informacoes(self):
        print(f"Nome: {self.nome}")
        print(f"Preço: {self.preco}")
        print(f"Quantidade em estoque: {self.quantidade_estoque}")

    def __str__(self):
        return f"Produto: {self.nome}, Preço: R${self.preco:.2f}, Quantidade em estoque: {self.quantidade_estoque}"

# Exemplo de uso:
if __name__ == "__main__":
    produto1 = Produto("Celular", 4000.0, 10)
    print(produto1)

    produto1.atualizar_preco(4500.0)
    produto1.atualizar_quantidade_estoque(8)
    print(produto1)
```
