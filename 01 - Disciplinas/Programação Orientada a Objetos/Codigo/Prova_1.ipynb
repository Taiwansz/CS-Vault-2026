```python
class Produto:
    def __init__(self, nome, codigo_barras, preco):
        self.nome = nome
        self.codigo_barras = codigo_barras
        self.preco = preco

    def calcular_preco(self, quantidade):
        return self.preco * quantidade

class Estoque:
    def __init__(self):
        self.produtos = {}

    def adicionar_produto(self, produto):
        if isinstance(produto, Produto):
            self.produtos[produto.codigo_barras] = produto
        else:
            print("Erro: O objeto não é um produto.")

    def remover_produto(self, codigo_barras):
        if codigo_barras in self.produtos:
            produto_removido = self.produtos.pop(codigo_barras)
            return f"Produto {produto_removido.nome} removido do estoque."
        return "Produto não encontrado no estoque."

class EstoqueAlimenticio(Estoque):
    def __init__(self):
        super().__init__()

class EstoqueHigiene(Estoque):
    def __init__(self):
        super().__init__()

class EstoqueLimpeza(Estoque):
    def __init__(self):
        super().__init__()

class Venda:
    def __init__(self):
        self.produtos_vendidos = []

    def adicionar_produto(self, produto):
        if isinstance(produto, Produto):
            self.produtos_vendidos.append(produto)
        else:
            print("Erro: O objeto não é um produto.")

    def remover_produto(self, codigo_barras):
        for produto in self.produtos_vendidos:
            if produto.codigo_barras == codigo_barras:
                self.produtos_vendidos.remove(produto)
                return f"Produto {produto.nome} removido da venda."
        return "Produto não encontrado na venda."


class ProdutoAlimenticio(Produto):
    def __init__(self, nome, codigo_barras, preco, data_validade):
        super().__init__(nome, codigo_barras, preco)
        self.data_validade = data_validade

class ProdutoHigiene(Produto):
    def __init__(self, nome, codigo_barras, preco, marca):
        super().__init__(nome, codigo_barras, preco)
        self.marca = marca

class ProdutoLimpeza(Produto):
    def __init__(self, nome, codigo_barras, preco, tipo):
        super().__init__(nome, codigo_barras, preco)
        self.tipo = tipo


estoque_alimenticio = EstoqueAlimenticio()
estoque_higiene = EstoqueHigiene()
estoque_limpeza = EstoqueLimpeza()

banana = ProdutoAlimenticio("Banana", "123456", 2.5, "2024-04-16")
coco = ProdutoAlimenticio("coco", "321654", 4.2, "2024-04-20")
sabonete = ProdutoHigiene("Sabonete", "789012", 1.99, "Dove")
shampoo = ProdutoHigiene("Shampoo", "841184", 10.85, "2027-02-25")
detergente = ProdutoLimpeza("Detergente", "456789", 3.5, "Líquido")
sabaoempo = ProdutoLimpeza("Sabão em Pó", "21897", 25.00, "2026-08-15")

estoque_alimenticio.adicionar_produto(banana)
estoque_alimenticio.adicionar_produto(coco)
estoque_higiene.adicionar_produto(sabonete)
estoque_higiene.adicionar_produto(shampoo)
estoque_limpeza.adicionar_produto(detergente)
estoque_limpeza.adicionar_produto(sabaoempo)

print("Produtos no estoque alimentício:", [produto.nome for produto in estoque_alimenticio.produtos.values()])
print("Produtos no estoque de higiene:", [produto.nome for produto in estoque_higiene.produtos.values()])
print("Produtos no estoque de limpeza:", [produto.nome for produto in estoque_limpeza.produtos.values()])
print("\n")
estoque_alimenticio.remover_produto("123456")
estoque_higiene.remover_produto("789012")
estoque_limpeza.remover_produto("456789")

print("Produtos no estoque alimentício após remover Banana:", [produto.nome for produto in estoque_alimenticio.produtos.values()])
print("Produtos no estoque de higiene após remover Sabonete:", [produto.nome for produto in estoque_higiene.produtos.values()])
print("Produtos no estoque de limpeza após remover Detergente:", [produto.nome for produto in estoque_limpeza.produtos.values()])
```
