# O Teste de Turing: Uma Simulação Interativa

Proposto por Alan Turing em 1950 no artigo *"Computing Machinery and Intelligence"*, o Teste de Turing (originalmente chamado de "Jogo da Imitação") avalia a capacidade de uma máquina de exibir comportamento inteligente equivalente, ou indistinguível, ao de um ser humano.

**Como funciona esta simulação:**
1. Você atuará como o **Juiz**.
2. Você fará perguntas para a **Entidade A** e para a **Entidade B**.
3. Uma entidade é um script de computador simples. A outra é uma tentativa de simular um humano respondendo (com atrasos e linguagem casual).
4. Seu objetivo é descobrir qual é o computador e qual é o "humano".

```python
import time
import random

def resposta_maquina(mensagem):
    msg = mensagem.lower()
    if "oi" in msg or "olá" in msg:
        return "opa, como posso ajudar?"
    elif "como" in msg and "você" in msg:
        return "tô funcionando de boa, valeu"
    elif "nome" in msg:
        return "sou só um bot, não tenho nome"
    elif "humano" in msg or "robô" in msg or "maquina" in msg:
        return "sou um programa pra tirar dúvidas"
    elif "tempo" in msg or "clima" in msg:
        return "tô sem acesso ao clima agora"
    else:
        respostas_padrao = [
            "entendi, manda mais",
            "massa, q q vc acha disso?",
            "não saquei, explica de novo?",
            "da hora isso"
        ]
        return random.choice(respostas_padrao)

def resposta_humano_simulado(mensagem):
    msg = mensagem.lower()

    time.sleep(random.uniform(1.5, 3.5))

    if "oi" in msg or "olá" in msg:
        return "opa, bão?"
    elif "como" in msg and "você" in msg:
        return "tô suave, só meio cansado e tu?"
    elif "nome" in msg:
        return "vou passar meu nome pra teste não mano"
    elif "humano" in msg or "robô" in msg:
        return "sou gente né, mó pergunta doida"
    elif "tempo" in msg or "clima" in msg:
        return "tá nublado aqui, deve chover logo"
    else:
        respostas_padrao = [
            "não peguei a ideia...",
            "pode crer, e aí?",
            "sei lá mano, nunca pensei nisso",
            "putz, difícil essa"
        ]
        return random.choice(respostas_padrao)

entidades = [resposta_maquina, resposta_humano_simulado]
random.shuffle(entidades)
entidade_a = entidades[0]
entidade_b = entidades[1]

print("pronto, tudo configurado pro teste")
```

```python
print("----------------------------------------")
print("        bora pro teste de turing        ")
print("----------------------------------------")
print("vc tem 3 chances pra falar com eles.")
print("pensa bem no q vai perguntar!\n")

rodadas = 3

for i in range(rodadas):
    print(f"\n--- rodada {i+1} de {rodadas} ---")

    pergunta_a = input("manda uma pra entidade A: ")
    print("A ta digitando...")
    resposta_a = entidade_a(pergunta_a)
    print(f">> A: {resposta_a}\n")

    pergunta_b = input("agora manda pro B: ")
    print("B ta digitando...")
    resposta_b = entidade_b(pergunta_b)
    print(f">> B: {resposta_b}\n")

print("----------------------------------------")
print("acabou o papo! hora da verdade.")
print("----------------------------------------\n")

palpite = input("quem vc acha q eh o robo? chuta ai (A ou B): ").strip().upper()

maquina_real = "A" if entidade_a == resposta_maquina else "B"

if palpite == maquina_real:
    print(f"\nboa! vc acertou. o robo era o {maquina_real}.")
    print("a maquina se deu mal dessa vez.")
else:
    print(f"\nixe, vc foi enganado! o robo era o {maquina_real}.")
    print("a maquina mandou bem e te enganou direitinho!")
```
